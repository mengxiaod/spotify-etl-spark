```python
import spotipy
from spotipy.oauth2 import SpotifyClientCredentials
import os
from dotenv import load_dotenv

load_dotenv()
```

```python
client_credentials_manager = SpotifyClientCredentials(client_id=os.getenv("SPOTIFY_CLIENT_ID"),
    client_secret=os.getenv("SPOTIFY_CLIENT_SECRET"))
```

```python
sp = spotipy.Spotify(client_credentials_manager = client_credentials_manager)
```

```python
playlist_link = "https://open.spotify.com/playlist/37i9dQZEVXbNG2KDcFcKOF?si=1333723a6eff4b7f&nd=1&dlsi=1586c2ede2dc4823"
```

```python
playlist_URI = playlist_link.split("/")[-1].split("?")[0]
playlist_URI
```

```python
#extrac data that stored in JSON file
data = sp.playlist_tracks(playlist_URI)
data
```

```python
#this dataset contains info about 50 songs (tracks), the info I need stored in 'item'
len(data['items'])
```

```python
# Let's check the first track:
data['items'][0]
```

### Parse JSON to get structured data of albumsm

```python
data['items'][0]['track']['album']['id']  #album_id
```

```python
data['items'][0]['track']['album']['name'] #album name
```

```python
data['items'][0]['track']['album']['release_date'] #release date
```

```python
data['items'][0]['track']['album']['total_tracks'] #total tracks
```

```python
data['items'][0]['track']['album']['external_urls']['spotify'] #album_url
```

```python
#structure the data for Album
album_list = []
for row in data['items']:
    album_id = row['track']['album']['id']
    album_name = row['track']['album']['name']
    album_release_date = row['track']['album']['release_date']
    album_total_tracks = row['track']['album']['total_tracks']
    album_url = row['track']['album']['external_urls']['spotify']
    album_element = {'album_id' : album_id,
                    'name' : album_name,
                    'release_date' : album_release_date,
                    'total_tracks' : album_total_tracks,
                    'url' : album_url}
    album_list.append(album_element)
```

```python
album_list
```

Parse JSON to get structured data of artists

```python
#similar to album, I also want to try the second method to extract artist data.
artist_list = []
for row in data['items']:
    for key, value in row.items():
        if key == 'track':
            for artist in value['artists']:
                artist_dict = {'artist_id' : artist['id'],
                'artist_name' : artist['name'],
                'external_url' : artist['href'],
                }
                artist_list.append(artist_dict)
```

```python
artist_list
```

### Parse JSON to get structured data of songs

```python
song_list = []
for row in data['items']:
    song_id = row['track']['id']
    song_name = row['track']['name']
    song_duration = row['track']['duration_ms']
    song_url = row['track']['external_urls']['spotify']
    song_popularity = row['track']['popularity']
    song_added = row['added_at']
    album_id = row['track']['album']['id']
    artist_id = row['track']['album']['artists'][0]['id']
    song_element = {'song_id':song_id,'song_name':song_name,'duration_ms':song_duration,'url':song_url,
                    'popularity':song_popularity,'song_added':song_added,'album_id':album_id,
                    'artist_id':artist_id
                   }
    song_list.append(song_element)

```

```python
song_list
```

### Convert to DataFrame

```python
import pandas as pd
```

```python
album_df = pd.DataFrame.from_dict(album_list)
#drop duplicates
album_df = album_df.drop_duplicates(subset = ['album_id'])
```

```python
artist_df = pd.DataFrame.from_dict(artist_list)
#drop duplicates
artist_df = artist_df.drop_duplicates(subset = ['artist_id'])
```

```python
song_df = pd.DataFrame.from_dict(song_list)
```

```python
#correct the datatype of 'release_date'
album_df['release_date'] = pd.to_datetime(album_df['release_date'])
album_df.info()
```

```python
#correct the datatype of 'song_added'
song_df['song_added'] = pd.to_datetime(song_df['song_added'])
song_df.info()
```
